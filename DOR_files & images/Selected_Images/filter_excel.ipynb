```python
import pandas as pd

# --- Step 1: Load the two CSV files ---
# Replace 'file1.csv' and 'file2.csv' with your actual file paths
file1 = 'Combined_Excel_Results/combined_MMDetection_0.2.csv'
file2 = 'Combined_Excel_Results/combined_MMdetection_cleaned.csv'

df1 = pd.read_csv(file1)
df2 = pd.read_csv(file2)

# --- Step 2: Define the common column ---
common_column = 'score'

# --- Step 3: Filter rows that have matching values in the common column ---
# 'inner' join keeps only matching rows
matching_rows = pd.merge(df1, df2, on=common_column, how='inner')

# --- Step 4: Save or display the result ---
matching_rows.to_csv('matching_rows.csv', index=False)

print("✅ Matching rows saved to 'matching_rows.csv'")
print(matching_rows.head())  # Show a preview
```
